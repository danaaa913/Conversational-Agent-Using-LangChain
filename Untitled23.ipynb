```python
# ============================================
# Chat with Your Context - Simple Version
# ============================================

from fastapi import FastAPI, UploadFile, File, HTTPException
from pydantic import BaseModel
from pypdf import PdfReader
from langchain_community.vectorstores import Chroma
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
import os
import json
from datetime import datetime

app = FastAPI(title="Context-Aware Conversational Agent")

# ============================================
# Configuration
# ============================================
VECTORDB_PATH = "chroma_db"
EMBEDDING_MODEL = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Simple memory store (dictionary بدل langchain memory)
conversations = {}

# ============================================
# Helper Functions
# ============================================

def extract_text_from_pdf(file_path: str) -> str:
    """استخراج النص من ملف PDF"""
    reader = PdfReader(file_path)
    text = ""
    for page in reader.pages:
        page_content = page.extract_text()
        if page_content:
            text += page_content + "\n"
    return text

def create_vector_store(text: str):
    """تقسيم النص وإنشاء vector store"""
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200,
        length_function=len
    )
    chunks = text_splitter.create_documents([text])

    vectorstore = Chroma.from_documents(
        documents=chunks,
        embedding=EMBEDDING_MODEL,
        persist_directory=VECTORDB_PATH
    )
    vectorstore.persist()
    return vectorstore, len(chunks)

# ============================================
# API Models
# ============================================

class ChatMessage(BaseModel):
    message: str
    session_id: str = "default"

class UploadResponse(BaseModel):
    status: str
    message: str
    chunks_created: int

class ChatResponse(BaseModel):
    answer: str
    sources_count: int
    session_id: str
    timestamp: str

# ============================================
# API Endpoints
# ============================================

@app.post("/upload", response_model=UploadResponse)
async def upload_document(file: UploadFile = File(...)):
    """رفع وثيقة PDF ومعالجتها"""
    if not file.filename.endswith('.pdf'):
        raise HTTPException(status_code=400, detail="فقط ملفات PDF مسموحة")

    temp_path = f"temp_{file.filename}"
    try:
        with open(temp_path, "wb") as f:
            content = await file.read()
            f.write(content)

        text = extract_text_from_pdf(temp_path)

        if not text.strip():
            raise HTTPException(status_code=400, detail="لم يتم استخراج نص من الملف")

        vectorstore, num_chunks = create_vector_store(text)

        return UploadResponse(
            status="success",
            message=f"✅ تم رفع ومعالجة الوثيقة بنجاح",
            chunks_created=num_chunks
        )

    finally:
        if os.path.exists(temp_path):
            os.remove(temp_path)

@app.post("/chat", response_model=ChatResponse)
async def chat_with_document(chat: ChatMessage):
    """المحادثة مع الوثيقة مع الحفاظ على السياق"""

    if not os.path.exists(VECTORDB_PATH):
        raise HTTPException(
            status_code=400,
            detail="❌ لم يتم رفع أي وثيقة بعد! استخدم /upload أولاً"
        )

    try:
        # إنشاء session إذا لم تكن موجودة
        if chat.session_id not in conversations:
            conversations[chat.session_id] = {
                "history": [],
                "context": ""
            }

        # تحميل vector store
        vectorstore = Chroma(
            persist_directory=VECTORDB_PATH,
            embedding_function=EMBEDDING_MODEL
        )

        # البحث في الوثيقة
        retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
        docs = retriever.get_relevant_documents(chat.message)

        # بناء الإجابة
        if docs:
            answer = "\n\n---\n\n".join([doc.page_content for doc in docs])
            sources_count = len(docs)
        else:
            answer = "❌ لم أجد معلومات ذات صلة في الوثيقة."
            sources_count = 0

        # حفظ في السياق
        session = conversations[chat.session_id]
        session["context"] = answer  # آخر سياق
        session["history"].append({
            "user_message": chat.message,
            "assistant_response": answer[:300] + "..." if len(answer) > 300 else answer,
            "timestamp": datetime.now().isoformat()
        })

        return ChatResponse(
            answer=answer,
            sources_count=sources_count,
            session_id=chat.session_id,
            timestamp=datetime.now().isoformat()
        )

    except Exception as e:
        raise HTTPException(status_code=500, detail=f"❌ خطأ: {str(e)}")

@app.get("/history/{session_id}")
async def get_conversation_history(session_id: str):
    """الحصول على تاريخ المحادثة"""
    if session_id not in conversations:
        return {
            "session_id": session_id,
            "message": "❌ لا يوجد محادثة لهذا الـ session",
            "history": []
        }

    return {
        "session_id": session_id,
        "history": conversations[session_id]["history"],
        "total_messages": len(conversations[session_id]["history"])
    }

@app.delete("/clear/{session_id}")
async def clear_conversation(session_id: str):
    """مسح تاريخ المحادثة"""
    if session_id in conversations:
        conversations[session_id]["history"].clear()
        conversations[session_id]["context"] = ""
        return {"message": f"✅ تم مسح المحادثة للـ session: {session_id}"}
    return {"message": "❌ Session غير موجود"}

@app.get("/")
async def root():
    return {
        "title": "🤖 Context-Aware Conversational Agent",
        "description": "Chat with your PDF documents while maintaining conversation context",
        "version": "1.0",
        "endpoints": {
            "POST /upload": "رفع ملف PDF",
            "POST /chat": "المحادثة مع الوثيقة",
            "GET /history/{session_id}": "عرض السجل",
            "DELETE /clear/{session_id}": "مسح السجل"
        }
    }

# ============================================
# Server Setup
# ============================================

import nest_asyncio
from pyngrok import ngrok
import uvicorn
from threading import Thread

nest_asyncio.apply()

try:
    public_url = ngrok.connect(8000)
    print("="*70)
    print("🎯 Context-Aware Conversational Agent - الجاهز للعمل")
    print("="*70)
    print(f"🌐 الرابط العام: {public_url}")
    print(f"📖 الوثائق التفاعلية: {public_url}/docs")
    print("="*70)
except:
    print("⚠️ ngrok غير متاح، السيرفر محلي فقط")
    print("📍 الوصول المحلي: http://localhost:8000/docs")

def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000, log_level="info")

thread = Thread(target=run_server, daemon=True)
thread.start()

print("✅ السيرفر شغال الآن!")
```

```python
# ============================================
# Chat with Your Context - Final Working Version
# ============================================

from fastapi import FastAPI, UploadFile, File, HTTPException
from pydantic import BaseModel
from pypdf import PdfReader
from langchain_community.vectorstores import Chroma
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langgraph.graph import StateGraph, END
from typing import Dict, List, TypedDict
import os
from datetime import datetime

app = FastAPI(title="Context-Aware Agent with LangGraph")

# ============================================
# Global Configuration
# ============================================
VECTORDB_PATH = "chroma_db_final"  # اسم مختلف
EMBEDDING_MODEL = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
sessions = {}

# ============================================
# State Definition for LangGraph
# ============================================

class AgentState(TypedDict):
    """تعريف حالة الـ Agent"""
    user_query: str
    documents: List[str]
    answer: str
    history: List[Dict]

# ============================================
# Helper Functions
# ============================================

def extract_pdf_text(file_path: str) -> str:
    """استخراج نص من PDF"""
    reader = PdfReader(file_path)
    text = ""
    for page in reader.pages:
        content = page.extract_text()
        if content:
            text += content + "\n"
    return text

def create_vector_db(text: str) -> int:
    """إنشاء قاعدة بيانات المتجهات"""
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200
    )
    docs = splitter.create_documents([text])
    vectorstore = Chroma.from_documents(
        documents=docs,
        embedding=EMBEDDING_MODEL,
        persist_directory=VECTORDB_PATH
    )
    vectorstore.persist()
    return len(docs)

# ============================================
# Graph Nodes (معالجات الحالة)
# ============================================

def retrieve_documents(state: AgentState) -> AgentState:
    """Node 1: استرجاع المستندات ذات الصلة"""
    print("📖 [Node 1] Retrieving relevant documents...")

    if not os.path.exists(VECTORDB_PATH):
        state["documents"] = []
        state["answer"] = "❌ لم يتم رفع أي وثيقة بعد"
        print("   ❌ No documents found in database")
        return state

    vectorstore = Chroma(
        persist_directory=VECTORDB_PATH,
        embedding_function=EMBEDDING_MODEL
    )

    retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
    docs = retriever.get_relevant_documents(state["user_query"])

    state["documents"] = [doc.page_content for doc in docs]
    print(f"   ✅ Found {len(docs)} relevant documents")

    return state

def generate_answer(state: AgentState) -> AgentState:
    """Node 2: توليد الإجابة من المستندات"""
    print("✍️ [Node 2] Generating answer...")

    if state["documents"]:
        state["answer"] = "\n\n---\n\n".join(state["documents"])
        print(f"   ✅ Answer generated ({len(state['answer'])} characters)")
    else:
        state["answer"] = "❌ لم أجد معلومات ذات صلة في الوثيقة"
        print("   ❌ No relevant information found")

    return state

def save_to_history(state: AgentState) -> AgentState:
    """Node 3: حفظ المحادثة في السجل"""
    print("💾 [Node 3] Saving to history...")

    if not state["history"]:
        state["history"] = []

    state["history"].append({
        "user_message": state["user_query"],
        "assistant_response": state["answer"][:300] + "..." if len(state["answer"]) > 300 else state["answer"],
        "timestamp": datetime.now().isoformat()
    })

    print(f"   ✅ Saved to history (total: {len(state['history'])} messages)")
    return state

# ============================================
# Build LangGraph
# ============================================

def create_agent_graph():
    """بناء رسم البيان للـ Agent"""

    # إنشاء رسم بياني جديد
    graph = StateGraph(AgentState)

    # إضافة الـ Nodes (المعالجات)
    graph.add_node("retrieve", retrieve_documents)
    graph.add_node("generate", generate_answer)
    graph.add_node("save_history", save_to_history)

    # إضافة الـ Edges (المسارات)
    graph.add_edge("retrieve", "generate")      # بعد استرجاع، توليد
    graph.add_edge("generate", "save_history")  # بعد توليد، حفظ
    graph.add_edge("save_history", END)         # نهاية

    # تعيين نقطة البداية
    graph.set_entry_point("retrieve")

    # ترجمة (compile) الرسم البياني
    return graph.compile()

# تهيئة الـ Graph عند بدء التطبيق
agent_graph = create_agent_graph()

# ============================================
# API Models (Pydantic Models)
# ============================================

class ChatRequest(BaseModel):
    query: str
    session_id: str = "default"

class ChatResponse(BaseModel):
    answer: str
    documents_count: int
    session_id: str
    message_count: int

# ============================================
# API Endpoints
# ============================================

@app.post("/upload")
async def upload_pdf(file: UploadFile = File(...)):
    """رفع ملف PDF ومعالجته"""

    if not file.filename.endswith('.pdf'):
        raise HTTPException(status_code=400, detail="فقط ملفات PDF مسموحة")

    temp_file = f"temp_{file.filename}"
    try:
        # حفظ الملف
        with open(temp_file, "wb") as f:
            f.write(await file.read())

        # استخراج النص
        text = extract_pdf_text(temp_file)

        if not text.strip():
            raise HTTPException(status_code=400, detail="لم يتم استخراج نص من الملف")

        # إنشاء قاعدة البيانات
        chunks_count = create_vector_db(text)

        return {
            "status": "✅ success",
            "message": f"تم رفع الملف بنجاح",
            "chunks_created": chunks_count,
            "filename": file.filename
        }

    finally:
        if os.path.exists(temp_file):
            os.remove(temp_file)

@app.post("/chat", response_model=ChatResponse)
async def chat_endpoint(request: ChatRequest):
    """المحادثة مع الـ Agent باستخدام LangGraph"""

    # إنشاء session جديد إذا لم يكن موجوداً
    if request.session_id not in sessions:
        sessions[request.session_id] = {"history": []}

    # تهيئة الحالة الأولية
    initial_state: AgentState = {
        "user_query": request.query,
        "documents": [],
        "answer": "",
        "history": sessions[request.session_id]["history"]
    }

    print(f"\n{'='*60}")
    print(f"🚀 Processing request for session: {request.session_id}")
    print(f"{'='*60}")

    # تشغيل الرسم البياني (Graph)
    result_state = agent_graph.invoke(initial_state)

    print(f"{'='*60}\n")

    # تحديث السجل
    sessions[request.session_id]["history"] = result_state["history"]

    return ChatResponse(
        answer=result_state["answer"],
        documents_count=len(result_state["documents"]),
        session_id=request.session_id,
        message_count=len(result_state["history"])
    )

@app.get("/history/{session_id}")
async def get_history(session_id: str):
    """الحصول على سجل المحادثة"""

    if session_id not in sessions:
        return {
            "session_id": session_id,
            "history": [],
            "message": "❌ No conversation history found"
        }

    return {
        "session_id": session_id,
        "history": sessions[session_id]["history"],
        "total_messages": len(sessions[session_id]["history"])
    }

@app.delete("/clear/{session_id}")
async def clear_history(session_id: str):
    """مسح سجل المحادثة"""

    if session_id in sessions:
        sessions[session_id]["history"].clear()
        return {"message": f"✅ تم مسح السجل للـ session: {session_id}"}

    return {"message": "❌ Session not found"}

@app.get("/graph-info")
async def graph_info():
    """معلومات عن الرسم البياني"""

    return {
        "title": "LangGraph Agent Structure",
        "nodes": [
            {
                "id": "retrieve",
                "name": "Retrieve Documents",
                "description": "استرجاع المستندات ذات الصلة"
            },
            {
                "id": "generate",
                "name": "Generate Answer",
                "description": "توليد الإجابة"
            },
            {
                "id": "save_history",
                "name": "Save to History",
                "description": "حفظ في السجل"
            }
        ],
        "edges": [
            "retrieve → generate",
            "generate → save_history",
            "save_history → END"
        ]
    }

@app.get("/")
async def root():
    return {
        "app_name": "🤖 Context-Aware Conversational Agent",
        "version": "2.0",
        "technology": "FastAPI + LangGraph + Chroma",
        "endpoints": {
            "POST /upload": "رفع ملف PDF",
            "POST /chat": "المحادثة مع الـ Agent",
            "GET /history/{session_id}": "عرض السجل",
            "DELETE /clear/{session_id}": "مسح السجل",
            "GET /graph-info": "معلومات الرسم البياني"
        }
    }

# ============================================
# Server Setup
# ============================================

import nest_asyncio
from pyngrok import ngrok
import uvicorn
from threading import Thread

nest_asyncio.apply()

# استخدام port مختلف
PORT = 9000

try:
    public_url = ngrok.connect(PORT)
    print("\n" + "="*70)
    print("✅ Context-Aware Agent with LangGraph - RUNNING")
    print("="*70)
    print(f"🌐 Public URL: {public_url}")
    print(f"📖 Interactive Docs: {public_url}/docs")
    print("="*70 + "\n")
except Exception as e:
    print(f"\n⚠️ ngrok unavailable: {str(e)[:50]}")
    print(f"📍 Local access: http://localhost:{PORT}/docs\n")

def run_server():
    uvicorn.run(app, host="0.0.0.0", port=PORT, log_level="info")

# تشغيل السيرفر في thread منفصل
server_thread = Thread(target=run_server, daemon=True)
server_thread.start()

print(f"✅ Server is running with LangGraph on port {PORT}!")
```

```python
# ============================================
# Simple Version بدون ngrok complications
# ============================================

from fastapi import FastAPI, UploadFile, File, HTTPException
from pydantic import BaseModel
from pypdf import PdfReader
from langchain_community.vectorstores import Chroma
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langgraph.graph import StateGraph, END
from typing import Dict, List, TypedDict
import os
from datetime import datetime

app = FastAPI(title="🤖 Context-Aware Agent")

VECTORDB_PATH = "chroma_db_v2"
EMBEDDING_MODEL = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
sessions = {}

# ============================================
# State
# ============================================

class AgentState(TypedDict):
    user_query: str
    documents: List[str]
    answer: str
    history: List[Dict]

# ============================================
# Nodes
# ============================================

def retrieve_documents(state: AgentState) -> AgentState:
    print("📖 Retrieving...")
    if not os.path.exists(VECTORDB_PATH):
        state["documents"] = []
        state["answer"] = "❌ No documents"
        return state

    vectorstore = Chroma(
        persist_directory=VECTORDB_PATH,
        embedding_function=EMBEDDING_MODEL
    )
    retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
    docs = retriever.get_relevant_documents(state["user_query"])
    state["documents"] = [doc.page_content for doc in docs]
    return state

def generate_answer(state: AgentState) -> AgentState:
    print("✍️ Generating...")
    state["answer"] = "\n\n---\n\n".join(state["documents"]) if state["documents"] else "❌ No info"
    return state

def save_history(state: AgentState) -> AgentState:
    print("💾 Saving...")
    if not state["history"]:
        state["history"] = []
    state["history"].append({
        "q": state["user_query"],
        "a": state["answer"][:200] + "...",
        "t": datetime.now().isoformat()
    })
    return state

# ============================================
# Graph
# ============================================

def build_graph():
    graph = StateGraph(AgentState)
    graph.add_node("retrieve", retrieve_documents)
    graph.add_node("generate", generate_answer)
    graph.add_node("save", save_history)
    graph.add_edge("retrieve", "generate")
    graph.add_edge("generate", "save")
    graph.add_edge("save", END)
    graph.set_entry_point("retrieve")
    return graph.compile()

graph = build_graph()

# ============================================
# API
# ============================================

class ChatRequest(BaseModel):
    query: str
    session_id: str = "default"

@app.post("/upload")
async def upload(file: UploadFile = File(...)):
    if not file.filename.endswith('.pdf'):
        raise HTTPException(status_code=400, detail="PDF only")

    temp = f"temp_{file.filename}"
    try:
        with open(temp, "wb") as f:
            f.write(await file.read())

        reader = PdfReader(temp)
        text = "".join([p.extract_text() or "" for p in reader.pages])

        splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
        docs = splitter.create_documents([text])
        vectorstore = Chroma.from_documents(
            documents=docs,
            embedding=EMBEDDING_MODEL,
            persist_directory=VECTORDB_PATH
        )
        vectorstore.persist()

        return {"status": "✅", "chunks": len(docs)}
    finally:
        if os.path.exists(temp):
            os.remove(temp)

@app.post("/chat")
async def chat(req: ChatRequest):
    if req.session_id not in sessions:
        sessions[req.session_id] = {"history": []}

    state = {
        "user_query": req.query,
        "documents": [],
        "answer": "",
        "history": sessions[req.session_id]["history"]
    }

    result = graph.invoke(state)
    sessions[req.session_id]["history"] = result["history"]

    return {
        "answer": result["answer"],
        "docs": len(result["documents"]),
        "session_id": req.session_id
    }

@app.get("/")
async def root():
    return {"app": "Context-Aware Agent with LangGraph ✅"}

# ============================================
# Server - بدون ngrok المعقدة
# ============================================

import nest_asyncio
import uvicorn
from threading import Thread

nest_asyncio.apply()

def run():
    uvicorn.run(app, host="127.0.0.1", port=8888, log_level="error")

Thread(target=run, daemon=True).start()

print("""
✅ Server is RUNNING!
📍 Open: http://localhost:8888/docs
""")
```

```python
!pip install -q gradio chromadb pypdf langgraph langchain-huggingface PyPDF2

import gradio as gr
from pypdf import PdfReader
from langchain_community.vectorstores import Chroma
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langgraph.graph import StateGraph, END
from typing import List, TypedDict
import os
import shutil

VECTORDB = "gradio_db_clean"
EMBEDDING = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
SESSIONS = {}

# تنظيف قاعدة البيانات القديمة
if os.path.exists(VECTORDB):
    shutil.rmtree(VECTORDB)

class State(TypedDict):
    query: str
    docs: List[str]
    answer: str
    history: List

def retrieve(s: State) -> State:
    """استرجاع المستندات"""
    if not os.path.exists(VECTORDB):
        s["docs"] = []
        return s

    try:
        db = Chroma(
            persist_directory=VECTORDB,
            embedding_function=EMBEDDING
        )
        # استخدم similarity_search بدل get_relevant_documents
        results = db.similarity_search(s["query"], k=3)
        s["docs"] = [doc.page_content for doc in results]
    except Exception as e:
        print(f"Error in retrieve: {e}")
        s["docs"] = []

    return s

def generate(s: State) -> State:
    """توليد الإجابة"""
    if s["docs"]:
        s["answer"] = "\n\n---\n\n".join(s["docs"])
    else:
        s["answer"] = "❌ لم أجد معلومات ذات صلة"
    return s

def save_hist(s: State) -> State:
    """حفظ في السجل"""
    if not s["history"]:
        s["history"] = []
    s["history"].append({
        "user": s["query"],
        "bot": s["answer"][:150] + "..."
    })
    return s

# بناء الـ Graph
g = StateGraph(State)
g.add_node("retrieve", retrieve)
g.add_node("generate", generate)
g.add_node("save", save_hist)
g.add_edge("retrieve", "generate")
g.add_edge("generate", "save")
g.add_edge("save", END)
g.set_entry_point("retrieve")
graph = g.compile()

def upload_pdf(pdf_file):
    """رفع وقراءة PDF"""
    try:
        reader = PdfReader(pdf_file.name)
        text = ""

        for page in reader.pages:
            extracted = page.extract_text()
            if extracted:
                text += extracted + "\n"

        if not text.strip():
            return "❌ لم يتم استخراج نص"

        # تقسيم
        splitter = RecursiveCharacterTextSplitter(
            chunk_size=800,
            chunk_overlap=150
        )
        docs = splitter.create_documents([text])

        # حفظ في Chroma
        print(f"📝 تقسيم إلى {len(docs)} مقطع...")
        Chroma.from_documents(
            documents=docs,
            embedding=EMBEDDING,
            persist_directory=VECTORDB
        ).persist()

        return f"✅ تم بنجاح!\n📊 عدد المقاطع: {len(docs)}"

    except Exception as e:
        return f"❌ خطأ: {str(e)[:80]}"

def chat(question):
    """المحادثة"""
    try:
        session_id = "user1"
        if session_id not in SESSIONS:
            SESSIONS[session_id] = []

        result = graph.invoke({
            "query": question,
            "docs": [],
            "answer": "",
            "history": SESSIONS[session_id]
        })

        SESSIONS[session_id] = result["history"]
        return result["answer"]

    except Exception as e:
        return f"❌ خطأ: {str(e)[:80]}"

def get_history():
    """عرض السجل"""
    session_id = "user1"
    if session_id not in SESSIONS or not SESSIONS[session_id]:
        return "❌ لا توجد محادثات بعد"

    text = ""
    for i, msg in enumerate(SESSIONS[session_id], 1):
        text += f"**السؤال {i}:** {msg['user']}\n"
        text += f"**الإجابة {i}:** {msg['bot']}\n\n"

    return text

# ============================================
# GRADIO UI
# ============================================

with gr.Blocks(title="Context-Aware Agent") as app:
    gr.Markdown("# 🤖 Context-Aware Conversational Agent")
    gr.Markdown("ارفع ملف PDF واسأل عنه باستخدام LangGraph")

    with gr.Tabs():
        # Tab 1: Upload
        with gr.Tab("📄 رفع PDF"):
            pdf = gr.File(label="اختر ملف PDF", file_types=[".pdf"])
            btn_up = gr.Button("🚀 رفع", size="lg", variant="primary")
            out_up = gr.Textbox(label="النتيجة", interactive=False, lines=3)
            btn_up.click(upload_pdf, pdf, out_up)

        # Tab 2: Chat
        with gr.Tab("💬 اسأل"):
            q = gr.Textbox(
                label="السؤال",
                placeholder="اسأل عن الملف...",
                lines=2
            )
            btn_chat = gr.Button("🔍 ابحث", size="lg", variant="primary")
            ans = gr.Textbox(label="الإجابة", interactive=False, lines=5)
            btn_chat.click(chat, q, ans)

        # Tab 3: History
        with gr.Tab("📋 السجل"):
            btn_hist = gr.Button("📖 عرض", size="lg", variant="primary")
            hist = gr.Markdown()
            btn_hist.click(get_history, outputs=hist)

app.launch(share=True)
```
